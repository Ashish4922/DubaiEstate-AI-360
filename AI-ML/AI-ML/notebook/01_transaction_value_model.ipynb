```python
import numpy as np
import pandas as pd

for name, data in [("Dataset 1", df1), ("Dataset 2", df2)]:
    print(f"\n===== {name}: AREA DIFFERENCES BY PROCEDURE =====")

    valid = (
        data["PROCEDURE_AREA"].notna()
        & data["ACTUAL_AREA"].notna()
    )

    comparison = data.loc[valid].copy()

    comparison["AREA_DIFFERENT"] = ~np.isclose(
        comparison["PROCEDURE_AREA"].to_numpy(),
        comparison["ACTUAL_AREA"].to_numpy(),
        rtol=1e-5,
        atol=1e-8
    )

    result = comparison.groupby("PROCEDURE_EN")["AREA_DIFFERENT"].agg(
        different_count="sum",
        total_count="count",
        different_fraction="mean"
    )

    result["different_percentage"] = (
        result["different_fraction"] * 100
    )

    print(result.sort_values("different_count", ascending=False))
```

```python
# Compare the two area columns across the dataset

area_check = df1[
    ["TRANSACTION_NUMBER", "PROP_TYPE_EN",
     "PROCEDURE_AREA", "ACTUAL_AREA"]
].copy()

area_check["AREA_DIFFERENCE"] = (
    area_check["PROCEDURE_AREA"] - area_check["ACTUAL_AREA"]
).abs()

area_check["AREA_MATCH"] = (
    area_check["PROCEDURE_AREA"].round(2)
    == area_check["ACTUAL_AREA"].round(2)
)

print("Total records:", len(area_check))

print(
    "Records where area values match:",
    area_check["AREA_MATCH"].sum()
)

print(
    "Records where area values differ:",
    (~area_check["AREA_MATCH"]).sum()
)

print(
    "Records missing either area:",
    area_check[["PROCEDURE_AREA", "ACTUAL_AREA"]]
    .isna().any(axis=1).sum()
)

print("\nLargest area differences:")

display(
    area_check.sort_values(
        "AREA_DIFFERENCE",
        ascending=False
    ).head(20)
)
```

```python
# AI-ML Day 1: Target and feature validation

target_column = "TRANS_VALUE"

candidate_features = [
    "AREA_EN",
    "PROP_TYPE_EN",
    "PROP_SB_TYPE_EN",
    "ACTUAL_AREA",
    "ROOMS_EN",
    "PARKING",
    "IS_OFFPLAN_EN",
    "IS_FREE_HOLD_EN",
    "USAGE_EN",
    "NEAREST_METRO_EN",
    "NEAREST_MALL_EN",
    "NEAREST_LANDMARK_EN",
    "PROJECT_EN"
]

columns_to_review = [
    "PROCEDURE_AREA",
    "PROCEDURE_EN",
    "MASTER_PROJECT_EN"
]

excluded_columns = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "GROUP_EN",
    "TOTAL_BUYER",
    "TOTAL_SELLER"
]

required_columns = (
    [target_column]
    + candidate_features
    + columns_to_review
    + excluded_columns
)

# Validate the column definitions against both datasets
for dataset_name, dataset in [
    ("Dataset 1", df1),
    ("Dataset 2", df2)
]:
    missing_columns = [
        column for column in required_columns
        if column not in dataset.columns
    ]

    print(f"\n{dataset_name}")
    print("-" * 40)
    print("Rows:", len(dataset))
    print("Target column exists:", target_column in dataset.columns)

    if missing_columns:
        print("Missing required columns:", missing_columns)
    else:
        print("All required columns exist.")

    if target_column in dataset.columns:
        target = dataset[target_column]

        print("Missing target values:", target.isna().sum())
        print("Zero or negative target values:", (target <= 0).sum())

print("\nNumber of candidate predictors:", len(candidate_features))
print("Target variable:", target_column)
```

```python
# AI-ML Day 1: Investigate transaction procedures

for dataset_name, dataset in [
    ("Dataset 1", df1),
    ("Dataset 2", df2)
]:
    print(f"\n{'=' * 70}")
    print(f"{dataset_name}: Procedure investigation")
    print("=" * 70)

    summary = (
        dataset.groupby("PROCEDURE_EN", dropna=False)
        .agg(
            transaction_count=("TRANS_VALUE", "size"),
            median_transaction_value=("TRANS_VALUE", "median"),
            mean_transaction_value=("TRANS_VALUE", "mean"),
            property_types=("PROP_TYPE_EN", "nunique"),
            property_subtypes=("PROP_SB_TYPE_EN", "nunique")
        )
        .sort_values("transaction_count", ascending=False)
    )

    print(summary.round(2).to_string())
```

```python
# AI-ML Day 1: Target and Feature Validation

target_column = "TRANS_VALUE"

candidate_features = [
    "AREA_EN",
    "PROP_TYPE_EN",
    "PROP_SB_TYPE_EN",
    "ACTUAL_AREA",
    "ROOMS_EN",
    "PARKING",
    "IS_OFFPLAN_EN",
    "IS_FREE_HOLD_EN",
    "USAGE_EN",
    "NEAREST_METRO_EN",
    "NEAREST_MALL_EN",
    "NEAREST_LANDMARK_EN",
    "PROJECT_EN"
]

excluded_columns = [
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "GROUP_EN",
    "TOTAL_BUYER",
    "TOTAL_SELLER",
    "MASTER_PROJECT_EN"
]

for dataset_name, dataset in [
    ("Dataset 1", df1),
    ("Dataset 2", df2)
]:
    print(f"\n{'=' * 50}")
    print(dataset_name)
    print("=" * 50)

    print("Total records:", len(dataset))

    # Validate target
    print("Target column exists:", target_column in dataset.columns)
    print("Missing target values:", dataset[target_column].isna().sum())
    print(
        "Zero or negative target values:",
        (dataset[target_column] <= 0).sum()
    )

    # Validate candidate features
    available_features = [
        col for col in candidate_features
        if col in dataset.columns
    ]

    missing_features = [
        col for col in candidate_features
        if col not in dataset.columns
    ]

    print("\nAvailable candidate features:", len(available_features))
    print(available_features)

    if missing_features:
        print("Missing feature columns:", missing_features)

    # Review excluded columns
    available_exclusions = [
        col for col in excluded_columns
        if col in dataset.columns
    ]

    print("\nColumns excluded or reserved for review:")
    print(available_exclusions)

print("\nDay 1 target and feature validation completed.")
```

```python
# Day 1: Provisional Feature Selection

target_column = "TRANS_VALUE"

selected_features = [
    "AREA_EN",
    "PROP_TYPE_EN",
    "PROP_SB_TYPE_EN",
    "ACTUAL_AREA",
    "ROOMS_EN",
    "PARKING",
    "IS_OFFPLAN_EN",
    "IS_FREE_HOLD_EN",
    "USAGE_EN",
    "PROJECT_EN"
]

excluded_features = [
    "NEAREST_METRO_EN",
    "NEAREST_MALL_EN",
    "NEAREST_LANDMARK_EN",
    "TRANSACTION_NUMBER",
    "INSTANCE_DATE",
    "GROUP_EN",
    "TOTAL_BUYER",
    "TOTAL_SELLER",
    "MASTER_PROJECT_EN"
]

print("Target variable:", target_column)

print("\nSelected candidate features:")
for feature in selected_features:
    print("-", feature)

print("\nExcluded or reserved features:")
for feature in excluded_features:
    print("-", feature)

print("\nNumber of candidate predictors:", len(selected_features))
```

```python
# Day 1: Final Feature Validation

for dataset_name, dataset in [
    ("Dataset 1", df1),
    ("Dataset 2", df2)
]:
    missing_features = [
        feature
        for feature in selected_features
        if feature not in dataset.columns
    ]

    print(f"\n{dataset_name}")
    print("Selected feature count:", len(selected_features))
    print("Missing feature columns:", missing_features)

    if not missing_features:
        print("PASS: All selected feature columns exist.")

print("\nDay 1 feature-selection validation completed.")
```

```python
# AI-ML Day 1: Comparable-Sales Candidate Audit

candidate_sale_procedures = [
    "Sale",
    "Sell - Pre registration"
]

for dataset_name, dataset in [
    ("Dataset 1", df1),
    ("Dataset 2", df2)
]:
    print(f"\n{'=' * 55}")
    print(f"{dataset_name}: Comparable-Sales Candidate Audit")
    print("=" * 55)

    procedure_counts = dataset["PROCEDURE_EN"].value_counts(
        dropna=False
    )

    print("Primary candidate procedures:")

    candidate_count = 0

    for procedure in candidate_sale_procedures:
        count = int(procedure_counts.get(procedure, 0))
        candidate_count += count
        print(f"{procedure}: {count:,}")

    print(f"\nTotal primary candidate records: {candidate_count:,}")
    print(
        "Percentage of dataset:",
        round(candidate_count / len(dataset) * 100, 2),
        "%"
    )

    print("\nOther procedure categories requiring review:")
    other_procedures = procedure_counts.drop(
        labels=[
            p for p in candidate_sale_procedures
            if p in procedure_counts.index
        ],
        errors="ignore"
    )

    print(other_procedures.to_string())

print("\nDay 1 comparable-sales audit completed.")
```
