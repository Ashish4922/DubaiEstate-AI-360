```python
# Day 3: Baseline preprocessing pipeline

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Target column
target = "TRANS_VALUE"

# Exclude target, date, and potential identifiers from predictors
excluded_columns = [
    target,
    "INSTANCE_DATE",
    "TRANSACTION_NUMBER",
]

excluded_columns = [
    col for col in excluded_columns if col in train_df.columns
]

# Use only columns available in the training dataset
feature_columns = [
    col for col in train_df.columns
    if col not in excluded_columns
]

# Identify numeric and categorical features
numeric_features = train_df[feature_columns].select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = train_df[feature_columns].select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)

# Numeric preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features),
])

# Separate features and target
X_train = train_df[feature_columns].copy()
y_train = train_df[target].copy()

X_val = val_df[feature_columns].copy()
y_val = val_df[target].copy()

X_test = test_df[feature_columns].copy()
y_test = test_df[target].copy()

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation and test without fitting again
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

print("\nPreprocessing completed.")
print("Training matrix:", X_train_processed.shape)
print("Validation matrix:", X_val_processed.shape)
print("Test matrix:", X_test_processed.shape)
```

# Day 6: Improved Regression Model (Ridge Regression)

## Objective
Train and evaluate an improved regularized regression model (**Ridge Regression**) to address high variance and multicollinearity observed in the Day 4 Linear Regression baseline.

### Key Methodology:
1. **Regularization ($L_2$ Penalty):** Unregularized Ordinary Least Squares (OLS) suffers from coefficient inflation and extreme prediction variance on high-dimensional one-hot encoded sparse feature spaces (12,840 features). Ridge regression introduces an $L_2$ penalty ($\alpha \sum w_j^2$) to stabilize weight estimation.
2. **Hyperparameter Selection on Validation Data:** Test candidate penalty parameters $\alpha \in [0.01, 0.1, 1.0, 10.0, 100.0]$ using training data only (`X_train_processed`, `y_train`) and select the optimal model using validation data only (`X_val_processed`, `y_val`).
3. **Strict Data Leakage Prevention:** The test dataset (`X_test_processed`, `y_test`) is strictly kept untouched and isolated during hyperparameter selection.
4. **Final Evaluation & Benchmark:** The chosen Ridge model is evaluated on the untouched test dataset and compared side-by-side with the Day 4 Linear Regression baseline.

```python
# Day 6: Ridge Regression Hyperparameter Tuning on Validation Set
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    median_absolute_error,
    r2_score
)

# 1. Define hyperparameter search grid for L2 regularization
alphas = [0.01, 0.1, 1.0, 10.0, 100.0]
val_tuning_records = []
fitted_ridge_models = {}

print("=" * 60)
print("DAY 6: RIDGE REGRESSION HYPERPARAMETER TUNING (VALIDATION SET)")
print("=" * 60)

# 2. Train each candidate Ridge model using training data only and evaluate on validation data only
for alpha in alphas:
    ridge_candidate = Ridge(alpha=alpha, random_state=42)
    ridge_candidate.fit(X_train_processed, y_train)
    
    # Validation prediction
    y_val_pred_candidate = ridge_candidate.predict(X_val_processed)
    
    val_mae = mean_absolute_error(y_val, y_val_pred_candidate)
    val_rmse = np.sqrt(mean_squared_error(y_val, y_val_pred_candidate))
    val_medae = median_absolute_error(y_val, y_val_pred_candidate)
    val_r2 = r2_score(y_val, y_val_pred_candidate)
    
    val_tuning_records.append({
        "alpha": alpha,
        "Validation MAE": val_mae,
        "Validation RMSE": val_rmse,
        "Validation Median Absolute Error": val_medae,
        "Validation R2 Score": val_r2
    })
    fitted_ridge_models[alpha] = ridge_candidate
    print(f"Alpha: {alpha:>6.2f} | Val MAE: {val_mae:,.2f} | Val RMSE: {val_rmse:,.2f} | Val MedAE: {val_medae:,.2f} | Val R2: {val_r2:.4f}")

# 3. Create DataFrame of validation tuning results
val_tuning_df = pd.DataFrame(val_tuning_records)

# 4. Select best alpha based strictly on lowest validation MAE
best_record = val_tuning_df.sort_values(by="Validation MAE", ascending=True).iloc[0]
best_alpha = float(best_record["alpha"])
best_ridge_model = fitted_ridge_models[best_alpha]

# 5. Record best model configuration and validation metrics
ridge_config = {
    "model_name": "Ridge Regression",
    "best_alpha": best_alpha,
    "random_state": 42,
    "validation_mae": best_record["Validation MAE"],
    "validation_rmse": best_record["Validation RMSE"],
    "validation_medae": best_record["Validation Median Absolute Error"],
    "validation_r2": best_record["Validation R2 Score"]
}

print("\n" + "=" * 60)
print(f"OPTIMAL HYPERPARAMETER SELECTED: alpha = {best_alpha}")
print(f"Best Validation MAE: {ridge_config['validation_mae']:,.2f}")
print(f"Best Validation RMSE: {ridge_config['validation_rmse']:,.2f}")
print(f"Best Validation R2 Score: {ridge_config['validation_r2']:.4f}")
print("=" * 60)

display(val_tuning_df)
```
